```python
# =====================================================================================
#  CELL 1  —  TRAIN ONCE, SAVE TO DISK.
#  Trains a learned diffusion (score) model for every (d, K), saves each checkpoint plus
#  the modes / schedule / R99 to  ./ckpt/ .  Skips anything already on disk, so you only
#  ever pay the training cost once.  Cell 2 loads these and never retrains.
# =====================================================================================
import os, math, json, time
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
import torch, torch.nn as nn, numpy as np
from scipy.stats import chi2

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

# ----------------------------- config -----------------------------
CKPT_DIR   = "./ckpt"
os.makedirs(CKPT_DIR, exist_ok=True)

sigma, Variance = 0.1, 0.1**2       # within-mode std / variance (data endpoint)
R_sphere        = 2.0               # modes live on a sphere of this radius
M_MULT          = 1.0               # min-separation multiplier
HALL_TARGET     = 0.03              # retrain (more steps) until learned hall-rate <= 3%
FORCE_RETRAIN   = False             # set True to ignore existing checkpoints

d_list = [2, 4, 8, 16, 32]
K_list = [2, 4, 8, 16]

def T_for(d):        return 150                          # sampling steps
def steps_for(d,K):  return int(3000*(1+d/16)*(1+K/16))  # SGD steps for training

# ----------------------------- schedule ---------------------------
def make_schedule(T):
    betas = np.linspace(0.0001, 0.02, T)
    return torch.tensor(np.cumprod(1 - betas), dtype=torch.float32, device=device)

# ----------------------------- score network ----------------------
class SinusoidalPosEmb(nn.Module):
    def __init__(self, dim): super().__init__(); self.dim = dim
    def forward(self, t):
        half = self.dim//2; f = math.log(10000)/(half-1)
        f = torch.exp(torch.arange(half, device=t.device)*-f)
        a = t[:,None].float()*f[None,:]
        return torch.cat([torch.sin(a), torch.cos(a)], 1)

class MLPBlock(nn.Module):
    def __init__(self, h, act):
        super().__init__(); self.n=nn.LayerNorm(h); self.a=act
        self.f1=nn.Linear(h,h); self.f2=nn.Linear(h,h)
    def forward(self, x):
        z=self.n(x); z=self.a(z); z=self.f1(z); z=self.a(z); z=self.f2(z)
        return x+z

class ScoreNet(nn.Module):
    def __init__(self, d, h=256, nb=4, td=128):
        super().__init__(); act=nn.LeakyReLU(0.2)
        self.inp=nn.Linear(d,h); self.te=SinusoidalPosEmb(td)
        self.tm=nn.Sequential(nn.Linear(td,h), act)
        self.bl=nn.ModuleList([MLPBlock(h,act) for _ in range(nb)])
        self.out=nn.Sequential(nn.LayerNorm(h), act, nn.Linear(h,d))
    def forward(self, x, t):
        z=self.inp(x)+self.tm(self.te(t))
        for b in self.bl: z=b(z)
        return self.out(z)

# ----------------------------- modes ------------------------------
def sample_modes(K, d, R, sigma, M_mult, seed=0):
    mult = 5.0 if d > 2 else 3.0
    min_sep2 = (mult*M_mult*sigma)**2
    rng = np.random.RandomState(seed); modes=[]; tries=0
    max_tries = 500*K + 2_000_000
    while len(modes) < K and tries < max_tries:
        v = rng.randn(d); v = R*v/np.linalg.norm(v)
        if all(np.dot(v-m, v-m) > min_sep2 for m in modes): modes.append(v)
        tries += 1
    return np.array(modes), mult

# ----------------------------- training ---------------------------
def train_learned(Means_t, d, K, abar, T, n_steps, lr=1e-3, batch=512, seed=0):
    torch.manual_seed(seed)
    sa=torch.sqrt(abar); soma=torch.sqrt(1-abar)
    m=ScoreNet(d).to(device); opt=torch.optim.Adam(m.parameters(), lr=lr)
    for _ in range(n_steps):
        k=torch.randint(0,K,(batch,),device=device)
        x0=Means_t[k]+sigma*torch.randn(batch,d,device=device)
        ti=torch.randint(0,T,(batch,),device=device); noise=torch.randn_like(x0)
        xt=sa[ti][:,None]*x0+soma[ti][:,None]*noise
        loss=((m(xt,ti)-noise)**2).mean()
        opt.zero_grad(); loss.backward(); opt.step()
    return m

@torch.no_grad()
def sample_learned(model, abar, T, N, d, seed, chunk=50000):
    """DDIM (deterministic) reverse pass. Returns (initial seeds X0, endpoints Xf)."""
    model.eval(); torch.manual_seed(seed)
    X=torch.randn(N,d,device=device); X0=X.clone()
    for i in reversed(range(1,T)):
        ab,abp=abar[i],abar[i-1]
        for s in range(0,N,chunk):
            xs=X[s:s+chunk]; ti=torch.full((xs.shape[0],),i,dtype=torch.long,device=device)
            eps=model(xs,ti); x0=(xs-torch.sqrt(1-ab)*eps)/torch.sqrt(ab)
            X[s:s+chunk]=torch.sqrt(abp)*x0+torch.sqrt(1-abp)*eps
    return X0, X

@torch.no_grad()
def label_fate(Xf, Means_t, R99, chunk=50000):
    N=Xf.shape[0]; lab=torch.empty(N,dtype=torch.long,device=device)
    for s in range(0,N,chunk):
        D=torch.cdist(Xf[s:s+chunk],Means_t); dmin,arg=D.min(1)
        lab[s:s+chunk]=torch.where(dmin<=R99,arg,torch.full_like(arg,-1))
    return lab

def ckpt_path(d,K): return os.path.join(CKPT_DIR, f"model_d{d}_K{K}.pt")

# ----------------------------- main loop --------------------------
manifest = {}
for d in d_list:
    R99 = float(np.sqrt(chi2.ppf(0.99,d))*sigma)
    for K in K_list:
        path = ckpt_path(d,K)
        if os.path.exists(path) and not FORCE_RETRAIN:
            print(f"d={d:>2} K={K:>2}  [cached] {path}")
            manifest[f"{d}_{K}"] = path; continue

        Means, mult = sample_modes(K,d,R_sphere,sigma,M_MULT)
        if len(Means) < K:
            print(f"d={d:>2} K={K:>2}: SKIP (placed {len(Means)}/{K})"); continue
        Means_t = torch.tensor(Means,dtype=torch.float32,device=device)
        T=T_for(d); abar=make_schedule(T); steps=steps_for(d,K)

        t0=time.time(); model=None; hr=1.0
        for attempt in range(6):
            model = train_learned(Means_t,d,K,abar,T,n_steps=steps)
            _,Xf_probe = sample_learned(model,abar,T,10000,d,seed=7)
            hr = (label_fate(Xf_probe,Means_t,R99)==-1).float().mean().item()
            if hr <= HALL_TARGET: break
            steps = int(steps*1.7)

        torch.save({
            "state_dict": model.state_dict(),
            "d": d, "K": K, "T": T,
            "means": Means_t.cpu(),
            "abar": abar.cpu(),
            "R99": R99, "sigma": sigma, "Variance": Variance,
            "hall_rate": hr, "mult": mult,
            "arch": {"h":256, "nb":4, "td":128},
        }, path)
        manifest[f"{d}_{K}"] = path
        print(f"d={d:>2} K={K:>2} mult={mult:.0f}Msigma T={T} hall={hr:.4f} "
              f"train={time.time()-t0:.1f}s -> saved {path}")

with open(os.path.join(CKPT_DIR,"manifest.json"),"w") as f:
    json.dump(manifest,f,indent=2)
print(f"\nCell 1 done. {len(manifest)} checkpoints in {CKPT_DIR}/  (manifest.json written).")
```
